```python
from pathlib import Path
from bs4 import BeautifulSoup as bs
import requests as req
from pprint import pprint
import json

html_path = Path("data4BS") / "2025 人工智慧基礎名詞大百科.html"
demoHTML_str = html_path.read_text(encoding="utf-8")

soup = bs(demoHTML_str, 'lxml')
```

### 🎯 CSS 選擇器精準定位：`.select_one()`
支援標準 **CSS Selector** 語法，能優雅解決多重 class 篩選與深層巢狀結構定位。
<br>
語法 `soup.select_one('selector')` 回傳**第一個**符合的 `Tag` 物件（或 `None`），比 `find()` 更具彈性。
<br>
這是處理複雜頁面結構的首選方案，可直接套用瀏覽器複製的 Selector 字串。


```python
# select_one()練習1
# 使用CSS選擇器 : 標籤選擇器
# 標籤選擇器 : 跟據指定標籤來選擇該元素
'''
練習對象
<a class="external-link" href="https://www.openai.com/" target="_blank">OpenAI 官網</a>
'''
type_a= soup.select_one('a')
print(f"標籤選擇器取得的a標籤元素:{type_a}")
```

```python
# select_one()練習2
# 使用CSS選擇器 : ID選擇器
# ID選擇器 : 根據指定id屬性值來選擇該元素
'''練習對象:
        <section id="chapter-2" class="section-block">
            <h2>第二章：學習方式與模型分類</h2>
            <a href="https://www.openai.com/" class="external-link" target="_blank">OpenAI 官網</a>
                        ...
'''
id_chapter2= soup.select_one('#chapter-2')
print(id_chapter2.name)
```

```python
# select_one()練習 3-1
# 使用CSS選擇器 : 類別選擇器
# 類別選擇器 : 根據類別(class)屬性來選擇該元素
'''練習對象:
        <a href="https://www.openai.com/" class="external-link" target="_blank">OpenAI 官網</a>
'''
pract_class_select= soup.select_one('.external-link')
print(pract_class_select.name)
```

```python
# select_one()練習 3-2
# 使用CSS選擇器 : 類別選擇器
#多重類別應用
#HTML 擁有多個 class 時，任一個都能選中
'''練習對象:<section id="chapter-3" class="section-block bg-gray">
            <h2>第三章：神經網路與技術細節</h2>
            <a href="https://arxiv.org/" class="external-link" target="_blank">ArXiv 論文網</a>

            <ul class="threads-list">
                <li class="post-item">'''
multi_class_select_1= soup.select_one('.section-block')
print(multi_class_select_1.name)
multi_class_select_2= soup.select_one('.bg-gray')
print(multi_class_select_2.name)

#select_one() 交集選擇，必須同時擁有兩個class才選中
multi_class_select_3= soup.select_one('.section-block.bg-gray')
print(multi_class_select_3.name)
```

```python
# select_one() 練習 4
# 使用CSS選擇器 : 屬性選擇器
# 根據特定屬性(ex:href)來選擇該元素，可提供限定條件，如指定屬性值
'''練習對象:
<a class="external-link" href="https://www.openai.com/" target="_blank">OpenAI 官網</a>

'''
attr_select= soup.select_one('[href]')
print(attr_select)

# 完全匹配 - 指定屬性值必須完全相同才能選中
attr_exact_select= soup.select_one('[href="https://www.openai.com/"]')
print(attr_exact_select)
```

```python
# select_one() 練習 5
# 複合選擇器 (Compound Selectors)
# 基本公式 : 標籤名 + 修飾符 (中間無空格，代表AND)

'''
<a class="external-link" href="https://www.openai.com/" target="_blank">OpenAI 官網</a>
'''
# Tag + Class
compound_tag_class= soup.select_one('a.external-link')
print(compound_tag_class)

# Tag + Attribute
compound_tag_attr= soup.select_one('a[href="https://www.openai.com/"]')
print(compound_tag_attr)

''' 
<section class="section-block bg-gray" id="chapter-3">
<h2>第三章：神經網路與技術細節</h2>
<a class="external-link" href="https://arxiv.org/" target="_blank">ArXiv 論文網</a>
<ul class="threads-list">
<li class="post-item">
<span class="hashtag">#架構</span>
<span class="content">Transformer</span>
<small>(現代大型語言模型（如BERT, GPT）的基礎架構。)</small>
</li>
<li class="post-item"><span class="tag">#單位</span> <span class="content">Token</span> <small>(模型處理文字的最小單位，不完全等於單字。)</small></li>
<li class="thread-post"><span class="tag">#參數</span> <span class="content">Parameter</span> <small>(模型內部可調整的變數，決定模型大小。)</small></li>
<li class="post-item"><span class="label">#參數</span> <span class="content">Weights</span> <small>(權重，神經網絡中連接強度的數值。)</small></li>
<li class="social-post"><span class="tag">#參數</span> <span class="content">Bias</span> <small>(偏差，神經元激活的閾值偏移量。)</small></li>
'''
# Tag + MultiClass
compound_tag_multiclass= soup.select_one('section.section-block.bg-gray')
print(compound_tag_multiclass)
print(compound_tag_multiclass['class'])

# Tag + Exact Attribute
compound_tag_exactattr= soup.select_one('a[href="https://www.openai.com/"]')
print(compound_tag_exactattr)
```

### 批次搜尋：`select()`
- **定義**：回傳**所有**符合 CSS 選擇器的元素 **列表 (List)**，找不到則回傳 `[]`。
- **操作**：列表不能直接取 `.text`，必須使用 **`for` 迴圈** 遍歷每個元素。

```python
# select() 練習
# 使用CSS選擇器定位HTML中的'li'，並且找class屬性包含'post-item'的所有元素
'''
練習對象:
<li class="post-item"><span class="tag">#單位</span> <span class="content">Token</span> <small>(模型處理文字的最小單位，不完全等於單字。)</small></li>
                <li class="thread-post"><span class="tag">#參數</span> <span class="content">Parameter</span> <small>(模型內部可調整的變數，決定模型大小。)</small></li>
                <li class="post-item"><span class="label">#參數</span> <span class="content">Weights</span> <small>(權重，神經網絡中連接強度的數值。)</small></li>
                <li class="social-post"><span class="tag">#參數</span> <span class="content">Bias</span> <small>(偏差，神經元激活的閾值偏移量。)</small></li>
'''
li_post_item = soup.select('li.post-item')
print(f"有{len(li_post_item)}個元素")
# 將元素內的文字取出，放入一個list中
text_list =[]
for string in li_post_item:
    text_list.append(string.text)
pprint(text_list)
```
