````python
from pathlib import Path
from bs4 import BeautifulSoup as bs
import requests as req
from pprint import pprint
import json

html_path = Path("data4BS") / "2025 人工智慧基礎名詞大百科.html"
demoHTML_str = html_path.read_text(encoding="utf-8")

soup = bs(demoHTML_str, 'lxml')
````

### 🔍 單點搜尋利器：`.find()`
此方法專用於鎖定 DOM 樹中「第一個」符合條件的節點，支援標籤名與屬性篩選。
不同於全數列出的 `.find_all()`，它僅回傳單一 `Tag` 物件或 `None`。
適合用於定位頁面標題、特定 ID 區塊或導航列等唯一性元素。


````python
# 使用find()方法取得標籤為<title>的元素
# find() 方法會回傳第一個符合條件的元素
````

````python
# 使用 .find() 的 string 參數，依照標籤中的「完整文字」進行搜尋
# 目標文字位於：<dd>物聯網，常與 AI 結合成為 AIoT。</dd>
# 注意：只指定 string 參數時，回傳的是 NavigableString 文字節點，而不是 <dd> 標籤
iot_string = soup.find(string='物聯網，常與 AI 結合成為 AIoT。')
print(iot_string)

# 進階用法 : 
# 方法一：從文字節點取得父標籤
iot_tag = iot_string.parent
print(iot_tag)

# 方法二：同時指定標籤名稱與文字
iot_tag = soup.find(
    'dd',
    string='物聯網，常與 AI 結合成為 AIoT。'
)
print(iot_tag)
````

### 🌲 全域搜尋：`find_all()`

> **`find_all()`** 是 BeautifulSoup 中用於遍歷 DOM 樹並檢索「所有」符合條件節點的核心方法。<br>它解決了 `find()` 僅能返回單一實例的限制，回傳的是一個包含所有匹配結果的列表 (`ResultSet`)。
